### e3_fastapi_pydantic_combined.py

```python
from pydantic import BaseModel
```

```python
class Movie(BaseModel):
    id: int
    title: str
    genre: str

watchlist: list[Movie] = []
```

There are two different things here:
1. Movie → Class (blueprint).
2. watchlist → Instance of Python's list class, currently an empty list.

When you add a Movie object

```python
movie1 = Movie(id=1, title="Inception", genre="Sci-Fi")

watchlist.append(movie1)
```

Now:
- watchlist is still a list instance.
- movie1 is an instance of the Movie class.
- The list contains a Movie instance.

```python
print(type(watchlist))
print(type(movie1))
```

```python
def update_movie(movie_id: int, updated_movie: Movie):
    print('watchlist is',watchlist)
    for index, movie in enumerate(watchlist):
        if movie.id == movie_id:
            watchlist[index] = updated_movie
            return updated_movie
    return {"error": "Movie not found"}
```

```python
watchlist = [{"id":1,"title":"Inception","genre":"Sci-Fi"},{"id":2,"title":"Spiderman","genre":"Sci-Fi"}]
```

```python
update_movie(1,watchlist)
```
