````python
import os

from dotenv import load_dotenv

load_dotenv()
````

````python
SERVICENOW_USERNAME=os.getenv("SERVICENOW_USERNAME")
SERVICENOW_PASSWORD=os.getenv("SERVICENOW_PASSWORD")
INSTANCE=os.getenv("SERVICENOW_INSTANCE")
````

````python
import httpx

# INSTANCE = "https://dev397466.service-now.com"
# SERVICENOW_USERNAME = "your_username"
# SERVICENOW_PASSWORD = "your_password"

url = f"https://{INSTANCE}/api/now/table/incident"

payload = {
    "short_description": "Test incident from Python with async",
    "description": "Created using ServiceNow REST API",
    "urgency": "2",
    "impact": "2",
}

async def create_incident():

    async with httpx.AsyncClient() as client:

        response = await client.post(
            url,
            auth=(SERVICENOW_USERNAME, SERVICENOW_PASSWORD),
            headers={
                "Accept": "application/json",
                "Content-Type": "application/json",
            },
            json=payload,
        )

        response.raise_for_status()

        print(response.json())


# if __name__ == "__main__":
#     asyncio.run(create_incident())
await create_incident()
````

## Error and troubleshooting
`RuntimeError: asyncio.run() cannot be called from a running event loop`
Sol: his happens because you are running the code inside a Jupyter Notebook.

Jupyter already has an active event loop. asyncio.run() tries to create and manage another event loop, which causes this error.

Replace:
```python
if __name__ == "__main__":
    asyncio.run(create_incident())
```
with
```python
await create_incident()
```
