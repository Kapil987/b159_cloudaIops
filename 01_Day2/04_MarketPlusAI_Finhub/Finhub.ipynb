### System Prompt and User Prompts

| Field | Meaning                    | Value      |
| ----- | -------------------------- | ---------- |
| `c`   | Current price              | 291.13     |
| `d`   | Change from previous close | -4.50      |
| `dp`  | Percentage change          | -1.5222%   |
| `h`   | Day's high price           | 297.14     |
| `l`   | Day's low price            | 289.62     |
| `o`   | Opening price              | 296.03     |
| `pc`  | Previous close price       | 295.63     |
| `t`   | Unix timestamp             | 1781294400 |


| Technique        | Example                                    |
| ---------------- | ------------------------------------------ |
| Zero-shot        | "Analyze this stock quote..."              |
| One-shot         | Give 1 example stock quote + analysis      |
| Few-shot         | Give 2-5 examples stock quotes + analyses  |
| Chain-of-Thought | "Think step by step before analyzing"      |
| Role Prompting   | "You are a senior equity research analyst" |

```python
import finnhub

client = finnhub.Client(api_key=userdata.get('FINNHUB_API_KEY'))

print(client.quote("AAPL"))
quote = client.quote("AAPL")

current_price = quote["c"]
change = quote["d"]
change_percent = quote["dp"]

print(f"AAPL: ${current_price}")
print(f"Change: {change} ({change_percent:.2f}%)")
```

```python
import finnhub
from groq import Groq
from google.colab import userdata

# Clients
finnhub_client = finnhub.Client(
    # api_key=os.environ["FINNHUB_API_KEY"]
    api_key=userdata.get('FINNHUB_API_KEY')
    )
groq_client = Groq(
    api_key=userdata.get('GROQ_API_KEY') )

# Get stock quote
quote = finnhub_client.quote("AAPL")

# is still Zero-Shot Prompting, because you are not giving any examples of expected input/output behavior.
# Zero-shot prompting with runtime context (context injection). The stock quote data is context, not a demonstration example
prompt = f"""
Analyze this stock quote:

Current Price: {quote['c']}
Day High: {quote['h']}
Day Low: {quote['l']}
Change %: {quote['dp']}
"""
## Ref https://console.groq.com/docs/text-chat
response = groq_client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "system", "content": "You are a stock market analyst."},
        {"role": "user", "content": prompt}
    ]
)

print(response.choices[0].message.content)
```
