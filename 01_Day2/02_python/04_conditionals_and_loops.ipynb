## Concept

`if`/`elif`/`else` run code only when a condition is true. `for` loops repeat over each item in something (like a list); `while` loops repeat as long as a condition stays true. **Classic gotcha**: `=` assigns a value, `==` checks if two things are equal — mixing them up is one of the most common beginner errors.

## Analogy

`if`/`elif`/`else` is a traveler at a fork checking signposts top to bottom
and walking down the FIRST one that matches, ignoring the rest. A `for` loop
is a train with a fixed, already-known list of stations — it visits each one
once and stops. A `while` loop is a car circling a roundabout until an exit
sign turns green — if nothing inside the loop ever makes that condition true,
the car circles forever (an infinite loop). And `=` vs `==` is the difference
between repainting the exit sign green versus just checking whether it's
already green.

## Basic syntax

Before the fuller example below, here's the bare syntax on its own.

````python
# if / elif / else
if condition:
    statements # mandatory thing abc

elif another_condition:
    statements # mandatory
else:
    statements # mandatory
````

````python
# if / elif / else
x = 5
if x > 10:
    print("big")
elif x > 0:
    print("positive")
else:
    print("zero or negative")

# for loop — over a simple list
for i in [1, 2, 3]:
    print(i)

# while loop — runs until the condition is False
n = 0
while n < 3:
    print(n)
    n = n + 1
````

## Worked example

````python
favorite_food = "sushi"
foods = ["pizza", "sushi", "tacos"]

for food in foods:
    if food == favorite_food:
        print(f"{food} is my favorite!")
    else:
        print(f"{food} is good too.")

# count = 0
# while count < 3:
#     print(f"count is {count}")
#     count = count + 1   # without this line, the loop never ends
````
