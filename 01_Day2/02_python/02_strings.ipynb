## Concept

**Strings** (`str`) are text. You can join them with `+`, but **f-strings** (`f"..."`) are the cleaner way to mix variables into text. Strings also have built-in **methods** like `.upper()`, `.strip()`, and `.split()` that transform or inspect them.

## Analogy

A messy string is like a wrinkled banner fresh off the printer — extra space
at the ends, careless lowercase lettering. `.strip()` irons out the wasted
space at the very ends (not the middle), `.upper()` reprints it in a
consistent case, and `.split()` is a pair of scissors cutting one long strip
into several smaller ones wherever a separator appears. None of these change
the *original* banner — each hands back a brand new strip, which is why you
can chain them: `raw.strip().upper()`.
