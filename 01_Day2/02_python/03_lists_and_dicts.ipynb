## Concept

A **list** (`[]`) holds an ordered sequence of values, accessed by position (index). A **dict** (`{}`) holds key-value pairs, accessed by key. Use a list when order/position matters; use a dict when you look things up by name.

## Analogy

A list is like a restaurant coat-check rack: you find your coat with a
numbered ticket (position), not by describing it. A dict is like a wall of
labeled mailboxes: you find a letter by the NAME on the box, not by which row
it's in. Two different lookup strategies for "a bunch of things in one
container" — pick a list when position is natural, a dict when a meaningful
name/key is natural.

## Worked example

````python
foods = ["pizza", "sushi", "tacos"]
print(foods[0])       # "pizza" — indexes start at 0

for food in foods:
    print(f"I like {food}")

friend_ages = {"Sam": 25, "Priya": 28}
print(friend_ages["Sam"])   # 25

for name, age in friend_ages.items():
    print(f"{name} is {age} years old")
````
