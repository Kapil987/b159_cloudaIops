## The class

Each favorite thing is one small object — data (`category`, `name`) bundled with behavior (`describe()`), same shape as `Friend` in notebook 6.

````python
class FavoriteThing:
    def __init__(self, category, name):
        self.category = category
        self.name = name

    def describe(self):
        return f"My favorite {self.category} is {self.name}."
````

## Storing entries in a list

Same list pattern as notebook 3 — but now each item is a `FavoriteThing` object instead of a plain string or dict.

````python
favorites = [
    FavoriteThing("food", "sushi"),
    FavoriteThing("movie", "The Matrix"),
    FavoriteThing("color", "blue"),
]
````

## A function that prints the summary

Same reuse-avoids-repetition idea as notebook 5 — one function, called once, loops over every entry (notebook 4's `for` loop) and uses each object's own `describe()` method.

````python
def print_summary(things):
    print("--- Your Favorite Things ---")
    for thing in things:
        print(thing.describe())


print_summary(favorites)
````
