## Concept

A **class** bundles related data *and* the functions (called **methods**) that act on that data, into one thing. Compare to notebook 3's dict: a dict just holds data (`{"Sam": 25}`); a class can hold that same data *and* behavior together (a `Friend` object that knows its own name/age, and can also `greet()`). `__init__` is the special method that runs when you create a new object, and `self` refers to "this specific object."

## Analogy

A dict is a plain index card: "Name: Sam, Age: 25" — just data, sitting
there, unable to do anything with itself. A class instance is a little robot
with that same data pinned to it (a name tag, an age stamp) that can also ACT
on it — it can introduce itself out loud. `class Friend:` is the blueprint;
`__init__` is the assembly step that stamps the name/age onto a brand new
robot; `self` just means "whichever specific robot is currently talking" —
`priya.greet()` and `sam.greet()` use the same blueprint but refer to
different robots.

## Worked example

General python == function
class in python == methods
self is a internel parameter, poitinger or internal variable

````python
class Friend:
    def __init__(self, name, age): # name = abc, age= 16
        self.name = name   # stored on this specific object
        self.age = age

    def greet(self):
        print(f"Hi, I'm {self.name} and I'm {self.age} years old.")


sam = Friend("Sam", 25) # sam is called as instance of class
sam.greet()

# greet()

priya = Friend("Priya", 28)
priya.greet()

# Friend.greet(sam)

# Compare: notebook 3's dict only held the data, with no built-in behavior
# friend_ages = {"Sam": 25, "Priya": 28}
````
