## Concept

A **function** (`def`) is a reusable, named block of code. **Parameters** are the inputs it takes; `return` sends a value back to whoever called it (different from `print`, which just displays something and gives nothing back to the code). Functions exist so you don't repeat the same logic everywhere.

## Analogy

A function with `return` is a vending machine: put coins in the right slots
(`a`, `b`), get something real back through the dispenser (`return a + b`)
that you can carry away and use elsewhere. A function that only `print`s is
the same machine with a megaphone instead of a dispenser — it announces
something out loud, but nothing physical comes out, so trying to grab "what
it gave you" gets you `None`.

## Basic syntax

Before the fuller example below, here's the bare syntax on its own.

````python
# A function with return — gives a value back
def greet(name):
    return f"Hello, {name}!"

message = greet("Ada")
print(message)

# A function with only print — gives nothing back
def shout(name):
    print(f"HELLO, {name}!")

nothing = shout("Ada")
print(nothing)   # None
````

## Worked example

````python
def print_food_report(foods, favorite):
    """Takes a list of foods and a favorite, prints a message for each."""
    for food in foods:
        if food == favorite:
            print(f"{food} is my favorite!")
        else:
            print(f"{food} is good too.")


def add(a, b):
    return a + b


print_food_report(["pizza", "sushi", "tacos"], "sushi")

result = add(3, 4)   # return gives us something to store
print(f"3 + 4 = {result}")
````
