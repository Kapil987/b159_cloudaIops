## Concept

`raise` deliberately triggers an exception yourself, instead of waiting for Python to hit an error naturally. It's how a function says "the input you gave me is invalid, and I refuse to continue" — the caller then either handles it with `try/except` (notebook 8) or the program crashes with a clear error message, which is often better than silently continuing with bad data.

## Analogy

`raise` is a security guard at a door who stops bad input right there and
says exactly what's wrong ("Age can't be negative, got -5"), instead of
silently shrugging it through and letting the problem surface somewhere else,
much later, with no clue where it came from. The guard doesn't decide what
happens next — that's the `try/except` booth's job, ready to catch whatever
gets thrown and decide whether to handle it gracefully or let it end the
show.

## Basic syntax

Before the fuller example below, here's the bare syntax on its own.

````python
# raise stops the function immediately with a clear error
def check_positive(x):
    if x < 0:
        raise ValueError(f"x can't be negative, got {x}")
    return x

# check_positive(-1)   # uncomment to see it crash with a clear message
print(check_positive(-5))
````

## Worked example

````python
def set_age(age):
    if age < 0:
        raise ValueError(f"Age can't be negative, got {age}")
    return age


print(set_age(30))

try:
    set_age(-5)
except ValueError as e:
    print(f"Caught it: {e}")
````
