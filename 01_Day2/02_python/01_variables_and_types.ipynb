## Concept

A **variable** is a name that stores a value. Python figures out the **type** of that value automatically — the four basics are `int` (whole numbers), `float` (decimal numbers), `str` (text), and `bool` (`True`/`False`). `print()` displays a value.

## Analogy

Think of a variable like a **labeled shipping crate**: the label on the outside
(`age`) is the variable name, and whatever's actually packed inside (`30`) is
the value. You never write "this crate is for numbers only" on the label —
whoever opens the crate just looks at what's inside and can tell right away
whether it's a number, a price, a scroll of text, or a simple on/off switch.
That's exactly how Python decides `int` vs. `float` vs. `str` vs. `bool` — by
looking at the value itself, not by anything you declared up front.
