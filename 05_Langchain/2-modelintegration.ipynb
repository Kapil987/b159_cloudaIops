### Streaming And Batch

### Streaming
Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses.
Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time:

```python
model.invoke("Write me a 200 words paragraph on how a bicycle stays balanced while moving")
```

```python
for chunk in model.stream("Write me a 200 words paragraph on how a bicycle stays balanced while moving"):
    print(chunk.text, end="|", flush=True)
```

```python
model.invoke("Why do bicycle gears make it easier to pedal uphill?")
```

```python
for chunk in model.stream("Why do bicycle gears make it easier to pedal uphill?"):
    print(chunk.text, end="|", flush=True)
```


#### Batch
Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
responses = model.batch([
    "Why do bicycle gears make it easier to pedal uphill?",
    "How do bicycle brakes convert motion into heat?",
    "Why are bicycle tires filled with air instead of being solid rubber?"
])
for response in responses:
    print(response)
```

```python
model.batch(
    ["Why do bicycle gears make it easier to pedal uphill?",
    "How do bicycle brakes convert motion into heat?",
    "Why are bicycle tires filled with air instead of being solid rubber?"],
    config={
        'max_concurrency': 5,  # Limit to 5 parallel calls
    }
)
```
